```python
%sql
CREATE OR REPLACE TABLE happiness.bronze.banco_mundial
COMMENT 'Bronze Banco Mundial (WDI): una fila por pais/entidad + indicador, formato ANCHO (anios como columnas, igual que el archivo fuente). 100% crudo: sin UNPIVOT, sin filtrar a los 26 indicadores elegidos, sin castear (ver D10/D14). El despivoteo y el filtro contra config_indicadores pasan a ser trabajo de Silver. Fuente: bulk oficial WDICSV.csv.'
TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
)
AS
SELECT *, current_timestamp() AS _ingest_ts, 'WDICSV.csv' AS _source
FROM read_files(
  '/Volumes/happiness/landing/archivos/WDI_CSV/WDICSV.csv',
  format => 'csv',
  header => true,
  inferColumnTypes => false
);

SELECT COUNT(*) AS filas, COUNT(DISTINCT `Indicator Code`) AS indicadores, COUNT(DISTINCT `Country Code`) AS entidades
FROM happiness.bronze.banco_mundial;
```
