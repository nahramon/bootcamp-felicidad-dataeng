```python
%sql
CREATE OR REPLACE TABLE happiness.bronze.undp
COMMENT 'Bronze UNDP HDR25 (Human Development Report): una fila por pais, formato ancho (~1.076 columnas, patron <indicador>_<anio>, ej. hdi_1990...hdi_2023). 100% crudo, columnas tal cual el CSV fuente, sin castear ni filtrar (ver D10). Despivoteo a formato largo queda para Silver. Encoding ISO-8859-1 (latin-1), no UTF-8.'
TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
)
AS
SELECT *, current_timestamp() AS _ingest_ts, 'HDR25_Composite_indices_complete_time_series.csv' AS _source
FROM read_files(
  '/Volumes/happiness/landing/archivos/UNDP HDR25/HDR25_Composite_indices_complete_time_series.csv',
  format => 'csv',
  header => true,
  encoding => 'ISO-8859-1',
  inferColumnTypes => false
);

SELECT COUNT(*) AS filas, COUNT(DISTINCT iso3) AS paises FROM happiness.bronze.undp;
```
