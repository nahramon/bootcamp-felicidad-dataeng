```python
%sql
CREATE OR REPLACE TABLE happiness.bronze.owid_cantril_ladder
COMMENT 'Bronze OWID - grafico happiness-cantril-ladder: satisfaccion con la vida autoreportada (escala Cantril), por pais y anio. 100% crudo, columnas tal cual el CSV fuente, sin castear ni filtrar (ver D10).'
TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
)
AS
SELECT *, current_timestamp() AS _ingest_ts, 'happiness-cantril-ladder.csv' AS _source
FROM read_files(
  '/Volumes/happiness/landing/archivos/OWID/happiness-cantril-ladder.csv',
  format => 'csv',
  header => true,
  inferColumnTypes => false
);

SELECT COUNT(*) AS filas FROM happiness.bronze.owid_cantril_ladder;
```

```python
%sql
CREATE OR REPLACE TABLE happiness.bronze.owid_gdp_happiness
COMMENT 'Bronze OWID - grafico gdp-vs-happiness: PIB per capita, satisfaccion con la vida y region OWID, por pais y anio. 100% crudo, columnas tal cual el CSV fuente, sin castear ni filtrar (ver D10).'
TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
)
AS
SELECT *, current_timestamp() AS _ingest_ts, 'gdp-vs-happiness.csv' AS _source
FROM read_files(
  '/Volumes/happiness/landing/archivos/OWID/gdp-vs-happiness.csv',
  format => 'csv',
  header => true,
  inferColumnTypes => false
);

SELECT COUNT(*) AS filas FROM happiness.bronze.owid_gdp_happiness;
```
